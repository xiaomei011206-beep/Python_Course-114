### 城市實例ch3_24.py : 從地球到月球約是3844400公里. 假設火箭的速度是一馬赫.

```python
# ch3_24.py
dist = 384400
speed = 1225
total_hours = dist // speed
days = total_hours //24
hours = total_hours % 24
print("總共需要天數")
print(days)
print("小時數")
print(hours)
```

### 程式實例ch4_15.py : 基本資料輸入與運算.

```python
# ch4_15.py
print("歡迎使用成績輸入系統")
name = input("請輸入姓名 : ")
engh = input("請輸入英文成績 : ")
math = input("請輸入數學成績 : ")
total = int(engh)+int(math)
print("%s 你的總分是 %d" % (name, total))
```

### 程式實例ch4_20.py : 請輸入華氏溫度 , 這個程式會輸出攝氏溫度.

```python
# ch4_20.py
f = float(input("請輸入華氏溫度 : "))
c = (f - 32) * 5/9
print("華氏 %4.1f 等於攝氏 %4.1f" % (f,c))
```

```python
# ch5_4.py
print("奇數偶數判斷")
num = input("請輸入任意正值 : ")
rem = int(num) % 2
if (rem == 0):
    print("%d 是偶數" % int(num))
else:
    print("%d 是奇數" % int(num))
```

### 練習: 判斷輸入的數字是否為3的倍數

```python
print("是否3的數判斷")
num = input("請輸入任意正值 : ")
rem = int(num) % 3
if (rem == 0):
    print("%d 是3的倍數" % int(num))
else:
    print("%d 不是3的倍數" % int(num))
```
